```python

class Solution(object):
    def twoSum(self, nums, target):
           lookup_dict = {}
           for index, value in enumerate(nums):
                result = target - value
                if lookup_dict.get(result) is None:
                    lookup_dict[value] = index
                else:
                    return [lookup_dict.get(result), index]



Solution().twoSum([3,2,4], 6)
```
