```python
def return_max_value(a, b):
    if a > b:
        return a
    else:
        return b
```

```python
array1 = np.array([[3,5,6]])
array2 = np.array([[5,1,7]])

return_max_value_vect = np.vectorize(return_max_value)

return_max_value_vect(array1, array2)
```
