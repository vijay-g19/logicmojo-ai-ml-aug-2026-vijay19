#### Question-1

Write a function `second_largest(numbers)` that takes a list of numbers and returns the second largest number in the list. Assume the list has at least two unique numbers.

Example:
```python
numbers = [10, 5, 20, 8, 15]
second = second_largest(numbers)
print(second)
# Expected output: 15
```

````python
import string

# Write your code here
numbers = [10, 5, 20, 8, 15]

def second_largest(number_list):
    first_largest_num = second_largest_num = float("-inf")
    for x in number_list:
        if first_largest_num < x:
            second_largest_num =  first_largest_num
            first_largest_num = x
        elif x > second_largest_num:
            second_largest_num = x

    return second_largest_num

second = second_largest(numbers)
print(second)




````

#### Question-2

Write a function `count_vowels(text)` that takes a string as input and returns the number of vowels (a, e, i, o, u - case-insensitive) present in the string.

Example:
```python
text = "Hello World"
vowel_count = count_vowels(text)
print(vowel_count)
# Expected output: 3
```


````python
# Write your code here
text = "Hello World"

def count_vowels(input_text):
    count = 0
    for x in input_text:
        if x.lower() in 'aeiou':
            count += 1
    return count

vowel_count = count_vowels(text)
print(vowel_count)
````

#### Question-4

Write a function `merge_sorted_lists(list1, list2)` that takes two sorted lists of integers, `list1` and `list2` and merges them into a single sorted list efficiently. One way is to simple concatenate the lists and sort. But think of a better approach.

Example:
```python
list1 = [1, 4, 5]
list2 = [1, 3, 4]
merged = merge_sorted_lists(list1, list2)
print(merged)
# Expected output: [1, 1, 3, 4, 4, 5]
```

````python
# Write your code here

def merge_sorted_lists(list1_input, list2_input):
    list1_input.extend(list2_input)
    return sorted(list1_input)

list1 = [1,4,5]
list2 = [1,3,4]

merged = merge_sorted_lists(list1, list2)

print(merged)

````

#### Question-6

Write a function `is_anagram(str1, str2)` that checks if two given strings are anagrams of each other (contain the same characters with the same frequencies, ignoring case).

Example:
```python
print(is_anagram("listen", "silent"))
print(is_anagram("hello", "world"))
# Expected output:
# True
# False
```

````python
# Write your code here

def is_anagram(str1, str2):
    if len(str1) !=  len(str2):
        return False

    str1_sorted = sorted(str1)
    str2_sorted = sorted(str2)

    for i in range(len(str1_sorted)):
        if str1_sorted[i] != str2_sorted[i]:
            return False
    return True

def is_anagram_v2(str1, str2):
    return sorted(str1) == sorted(str2)

print(is_anagram_v2('listen', 'silent'))
````

#### Question-7

Write a function `rotate_list(lst, k)` that rotates a given list lst to the right by `k` positions in-place. Assume `k` is a non-negative integer and can be greater than the length of the list.

Example:

```python
my_list = [1, 2, 3, 4, 5]
rotate_list(my_list, 2)
print(my_list)
# Expected output: [4, 5, 1, 2, 3]
```

````python
# Write your code here
def rotate_list(num_list, position):
    assert 0 <= position < len(num_list) , "position should be positive int and less than length of list"
    result_list = num_list[position+1::] + num_list[:position+1]
    num_list[:] = result_list

my_list = [1,2,3,4,5]
rotate_list(my_list, 2)
print(my_list)
````

#### Question-8

You have a list of dictionaries, where each dictionary represents a product with keys "name", "price", and "category". Write a function group_by_category(products) that takes this list and returns a dictionary where the keys are the unique categories and the values are lists of product names belonging to that category.

Example:

```python
products = [
    {"name": "Laptop", "price": 1200, "category": "Electronics"},
    {"name": "Book", "price": 25, "category": "Books"},
    {"name": "Tablet", "price": 300, "category": "Electronics"},
    {"name": "Fiction", "price": 15, "category": "Books"},
]
grouped_products = group_by_category(products)
print(grouped_products)
# Expected output: {'Electronics': ['Laptop', 'Tablet'], 'Books': ['Book', 'Fiction']}

```


````python
# Write your code here

def group_by_category(products_list):
    grouped_by_category = {}
    for product_dict in products_list:
        name = product_dict.get("name")
        category = product_dict.get("category")
        if category in grouped_by_category:
            grouped_by_category[category].append(name)
        else:
            grouped_by_category[category] = [name]
    return grouped_by_category

products = [
    {"name": "Laptop", "price": 1200, "category": "Electronics"},
    {"name": "Book", "price": 25, "category": "Books"},
    {"name": "Tablet", "price": 300, "category": "Electronics"},
    {"name": "Fiction", "price": 15, "category": "Books"},
]

grouped = group_by_category(products)

print(grouped)
````

#### Question-9

Write a Python program that prints the first `n` rows of Pascal's triangle. `n` is a positive integer and an user input. To know about Pascal's triangle see the [wikipedia article](https://en.wikipedia.org/wiki/Pascal%27s_triangle): 

Example:

```python
n = 5
# Expected output:
#     1
#    1 1
#   1 2 1
#  1 3 3 1
# 1 4 6 4 1
```

````python
# Write your code here

def build_pascal_triangle(n):
    triangle_list = []
    for i in range(n):
        if i == 0:
            triangle_list.append([1])
            continue
        elif i == 1:
            triangle_list.append([1, 1])
            continue
        internal_list = []
        for j in range(i+1):
            if j == 0:
                internal_list.append(1)
            elif j == i:
                internal_list.append(1)
            else:
                previous_list = triangle_list[i-1]
                internal_list.append(previous_list[j-1] + previous_list[j])
        triangle_list.append(internal_list)
    return triangle_list

def print_pascal_triangle(n):
    triangle = build_pascal_triangle(n)
    for i in range(n):
        internal_list = triangle[i]
        print( (n-i+1) * " "+" ".join(map(str, internal_list)))

print_pascal_triangle(5)


````

#### Question-10

Write a function `reverse_the_words(s)` which will take a string `s` as input and reverses the words in the string. 

Example:

```python
s = "Python is awesome and challenging"
r = reverse_the_words(s)
print(r)
# Expected output: challenging and awesome is Python
```

````python
# Write your code here

def reverse_the_words(str_text):
    str_list = str_text.split()
    str_reverse = " ".join(str_list[::-1])
    return str_reverse

s = "Python is awesome and challenging"
print(reverse_the_words(s))

````

#### Bonus question

Write a function `word_counter(s)` which will take a string `s` as input and returns a dictionary with count of words (case insensitive). Remove all the punctuations and special symbols present in the string.  

Example:
```python
s = "This city is a beautiful city. This is where I live"
d = word_counter(s)

# Expected output (output can be in any order): {"this": 2, "city": 2, "is": 2, "a": 1, "beautiful": 1, "where": 1, "i": 1, "live": 1}
```

````python
# Write your code here

def word_counter(text):
    cleaned_text = text.translate(str.maketrans('', '', string.punctuation))
    str_list = cleaned_text.lower().split()
    result_dict = {}
    for item in str_list:
        if item in result_dict:
            result_dict[item] += 1
        else:
            result_dict[item] = 1
    return result_dict

s = "This city is a beautiful city. This is where I live"
d = word_counter(s)
print(d)
````
